## Descrição
Nessa camada Gold são feitas junções das tabelas ind_prod_ihas com as tabelas bases e ilhas, enriquecendo os dados com novos campos.

```python
spark.sql("USE CATALOG mvp")
spark.sql("USE SCHEMA gold")
```

```python
%sql
SELECT ind.* FROM mvp.silver.ind_prod_ilha as ind;
```

```python
%sql
-- Cria uma nova tabela acrescentando o nome da base e da ilha
CREATE OR REPLACE TABLE ind_prod_ilha_completa AS 
SELECT base.Nome As NomeBase, 
  ilha.Nome As NomeIlha,
  ind.*
FROM mvp.silver.ind_prod_ilha as ind
INNER JOIN mvp.silver.ilhas as ilha
  ON ind.IdIlha = ilha.IdIlha
INNER JOIN mvp.silver.bases as base
  ON ind.IdBase = base.IdBase
```

```python
spark.sql("""
  COMMENT ON TABLE ind_prod_ilha_completa IS 
  'Contém informação sobre os indicadores de disponibilidade, performance e qualidade das ilhas das bases, incluindo o nome das ilhas.'
""")
```

```python
%sql
SELECT *
FROM ind_prod_ilha_completa
LIMIT 10
```
